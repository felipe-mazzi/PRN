```python
from pynq.overlays.base import BaseOverlay

base = BaseOverlay('base.bit')
base.init_rf_clks()

# Find the RF Data Converter IP in this overlay
rfdc_names = [k for k in base.ip_dict if 'rf' in k.lower() and 'data' in k.lower() or 'usp_rf' in k.lower()]
print("RFdc candidates:", rfdc_names)
print("All IP:", list(base.ip_dict.keys()))
```

```python
rfdc = base.radio.rfdc

def show_block(kind, t, b):
    tiles = rfdc.dac_tiles if kind == 'DAC' else rfdc.adc_tiles
    blk = tiles[t].blocks[b]
    print(f"--- {kind} tile {t} block {b} ---")
    for attr in ['BlockStatus', 'NyquistZone', 'MixerSettings']:
        try:
            print(f"  {attr}: {getattr(blk, attr)}")
        except Exception as e:
            print(f"  {attr}: n/a ({e})")

for t, b in [(0, 0), (2, 0)]:
    show_block('DAC', t, b)
for t, b in [(0, 0), (0, 1), (2, 0), (2, 1)]:
    show_block('ADC', t, b)
```
