# A tiny pseudo-random number generator

Python's `random` module generates pseudo-random values: they look random for everyday uses, but come from a repeatable calculation. Run the cells from top to bottom with **Shift+Enter**.

```python
import random

# A seed makes this example repeatable. Change it to get a different sequence.
random.seed(7)
```

## Generate some values

Each call to `randint(1, 100)` returns an integer from 1 through 100, including both endpoints. Run this cell again to see the same sequence; change the seed above for another sequence.

```python
numbers = [random.randint(1, 100) for _ in range(5)]
print(numbers)
```

The seed is useful when you want to reproduce an example. For passwords, security tokens, or other security-sensitive work, use Python's `secrets` module instead of `random`.
